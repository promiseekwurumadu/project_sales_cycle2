# Cycle 2 — Profitability & Loss Driver Analysis

This notebook analyses business profitability using a star schema built in Cycle 2.

```python
import pandas as pd
```

```python
category_margin = pd.read_csv("../data/processed/category_margin.csv")
loss_products = pd.read_csv("../data/processed/loss_products_by_category.csv")
loss_customers = pd.read_csv("../data/processed/loss_customers_summary.csv")
loss_customer_category = pd.read_csv("../data/processed/category_profit_loss_customers.csv")
discount_band = pd.read_csv("../data/processed/discount_band_margin.csv")
```

## Category Performance

This shows overall profitability by category.

```python
category_margin
```

Technology and Office Supplies show strong margins (~17%).

Furniture generates high revenue but has very low margin (~2.5%), indicating structural inefficiency.

## Loss-Making Products

This shows how many products lose money and where they are concentrated.

```python
loss_products
```

Loss-making products are concentrated in Furniture, confirming it as a structurally weak category.

## Loss-Making Customers

```python
loss_customers
```

155 customers are unprofitable, contributing to a total loss of approximately -71,224.

## Category Performance for Loss Customers

```python
loss_customer_category
```

All categories become unprofitable when sold to loss-making customers, indicating customer-level behaviour (discounting) is a key issue.

## Discount Impact on Profitability

```python
discount_band
```

Higher discount levels significantly reduce margins and often lead to losses.

## Final Insight

- Technology is the main profit driver  
- Office Supplies is stable  
- Furniture has weak margins  
- Losses are driven by:
  - specific products  
  - specific customers  
  - excessive discounting  

### Recommendations

- Limit discount levels  
- Review Furniture pricing and costs  
- Monitor customer profitability
